```python
import pandas as pd
import matplotlib.pyplot as plt

# Read data from the CSV file
df = pd.read_csv('digital_currency.csv')

# 1a. Open USD price graph for the entire period
df['Unnamed: 0'] = pd.to_datetime(df['Unnamed: 0'])
plt.plot(df['Unnamed: 0'], df[['open_USD']])
plt.title('Open USD Price for the Entire Period')
plt.xlabel('Date')
plt.ylabel('Open USD Price')
plt.show()

# 1b. Open USD price graph for the year 2019
df_2019 = df[(df['Unnamed: 0'] >= '2019-01-01') & (df['Unnamed: 0'] <= '2019-12-31')]
plt.plot(df_2019['Unnamed: 0'], df_2019[['open_USD']])
plt.title('Open USD Price for the Year 2019')
plt.xlabel('Date')
plt.ylabel('Open USD Price')
plt.show()

# 1c. Open USD price graph for December 2020
df_dec_2020 = df[(df['Unnamed: 0'] >= '2020-12-01') & (df['Unnamed: 0'] <= '2020-12-31')]
plt.plot(df_dec_2020['Unnamed: 0'], df_dec_2020[['open_USD']])
plt.title('Open USD Price for December 2020')
plt.xlabel('Date')
plt.ylabel('Open USD Price')
plt.show()

# 1d. Open USD price graph for October 2018 - July 2020
df_oct_2018_jul_2020 = df[(df['Unnamed: 0'] >= '2018-10-01') & (df['Unnamed: 0'] <= '2020-07-31')]
plt.plot(df_oct_2018_jul_2020['Unnamed: 0'], df_oct_2018_jul_2020[['open_USD']])
plt.title('Open USD Price for October 2018 - July 2020')
plt.xlabel('Date')
plt.ylabel('Open USD Price')
plt.show()

# 1e. Open USD price graph for the last two weeks of September 2020
df_sep_2020_last_two_weeks = df[(df['Unnamed: 0'] >= '2020-09-17') & (df['Unnamed: 0'] <= '2020-09-30')]
plt.plot(df_sep_2020_last_two_weeks['Unnamed: 0'], df_sep_2020_last_two_weeks[['open_USD']])
plt.title('Open USD Price for the Last Two Weeks of September 2020')
plt.xlabel('Date')
plt.ylabel('Open USD Price')
plt.show()

# 2a. Find the average low USD price for 2019
average_low_2019 = df_2019['low_USD'].mean()
print(f'Average Low USD Price for 2019: {average_low_2019}')

# 2b. Find the average low USD price for each month
df['month'] = df['Unnamed: 0'].dt.month
average_low_monthly = df.groupby('month')['low_USD'].mean()
print('Average Low USD Price for Each Month:')
print(average_low_monthly)

# 2c. Find the average low USD price for every spring week of 2020
df_spring_2020 = df[(df['Unnamed: 0'] >= '2020-03-01') & (df['Unnamed: 0'] <= '2020-05-31')]
average_low_spring_2020 = df_spring_2020.groupby(df_spring_2020['Unnamed: 0'].dt.isocalendar().week)['low_USD'].mean()
print('Average Low USD Price for Every Spring Week of 2020:')
print(average_low_spring_2020)
```
